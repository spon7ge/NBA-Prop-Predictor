```python
from sklearn.metrics import r2_score
from catboost import CatBoostRegressor, Pool
import numpy as np
import pandas as pd

def build_recent_weights(df, player_col, recent_n=40, recent_weight=3.0):
    w = np.ones(len(df), dtype=float)
    idx = df.groupby(player_col, sort=False).tail(recent_n).index
    w[idx] = recent_weight
    return w

def clean_feature_dtypes(X, cat_cols):
    Xc = X.copy()
    for c in Xc.columns:
        if c in cat_cols:
            continue
        if Xc[c].dtype == "bool":
            Xc[c] = Xc[c].astype(int)
        elif not np.issubdtype(Xc[c].dtype, np.number):
            Xc[c] = pd.to_numeric(Xc[c], errors="coerce")
    return Xc

def cat_params(base=None, use_gpu=False):
    p = dict(
        loss_function="RMSE",
        eval_metric="RMSE",
        iterations=3000,
        learning_rate=0.01,
        depth=8,
        l2_leaf_reg=10.0,
        min_data_in_leaf=50,
        random_strength=1.0,
        bootstrap_type="Bayesian",
        bagging_temperature=1.0,
        colsample_bylevel=0.7,
        grow_policy="SymmetricTree",
        leaf_estimation_method="Newton",
        random_seed=42,
        verbose=False,
    )
    if base:
        p.update(base)
    if use_gpu:
        p["task_type"] = "GPU"
    return p

def fit_train_val(
    train_df, val_df,
    features, target_col, date_col, player_col,
    recent_n=40, recent_weight=3.0,
    params=None, use_gpu=False
):
    # sort by date to ensure chronological order
    train_df = train_df.sort_values(date_col).reset_index(drop=True)
    val_df = val_df.sort_values(date_col).reset_index(drop=True)

    X_tr = train_df[features]
    y_tr = train_df[target_col].to_numpy()
    X_va = val_df[features]
    y_va = val_df[target_col].to_numpy()

    # detect categorical columns
    cat_cols = [c for c in features if train_df[c].dtype == "object" or "ID" in c]
    cat_idx = [X_tr.columns.get_loc(c) for c in cat_cols]

    # clean types
    X_tr = clean_feature_dtypes(X_tr, set(cat_cols))
    X_va = clean_feature_dtypes(X_va, set(cat_cols))

    # sample weights
    w_tr = build_recent_weights(train_df, player_col, recent_n, recent_weight)

    train_pool = Pool(X_tr, y_tr, cat_features=cat_idx, weight=w_tr)
    valid_pool = Pool(X_va, y_va, cat_features=cat_idx)

    p = cat_params(params, use_gpu=use_gpu)
    p.update(dict(use_best_model=True, early_stopping_rounds=200))
    model = CatBoostRegressor(**p)
    model.fit(train_pool, eval_set=valid_pool)

    best_iter = model.get_best_iteration() or p["iterations"]

    # retrain final model on train+val
    p_final = cat_params(params, use_gpu=use_gpu)
    p_final["iterations"] = int(best_iter)
    full_df = pd.concat([train_df, val_df], ignore_index=True)
    X_full = clean_feature_dtypes(full_df[features], set(cat_cols))
    y_full = full_df[target_col].to_numpy()
    w_full = np.concatenate([
        build_recent_weights(train_df, player_col, recent_n, recent_weight),
        np.ones(len(val_df))
    ])
    full_pool = Pool(X_full, y_full, cat_features=cat_idx, weight=w_full)

    final_model = CatBoostRegressor(**p_final)
    final_model.fit(full_pool, verbose=False)

    # validation metrics
    val_preds = model.predict(X_va)
    diff = val_preds - y_va
    val_rmse = float(np.sqrt(np.mean(diff**2)))
    val_mae = float(np.mean(np.abs(diff)))
    val_r2 = float(r2_score(y_va, val_preds))

    return final_model, dict(RMSE=val_rmse, MAE=val_mae, R2=val_r2, best_iteration=int(best_iter)), cat_cols

def evaluate_test(model, test_df, features, target_col, cat_cols):
    X_te = clean_feature_dtypes(test_df[features], set(cat_cols))
    y_te = test_df[target_col].to_numpy()
    preds = model.predict(X_te)
    diff = preds - y_te

    rmse = float(np.sqrt(np.mean(diff**2)))
    mae = float(np.mean(np.abs(diff)))
    r2 = float(r2_score(y_te, preds))

    residuals_df = pd.DataFrame({
        "actual": y_te,
        "predicted": preds,
        "residual": diff
    })
    
    # Join back some useful context for grouping
    context_cols = ["PLAYER_ID", "TEAM_ID", "OPP_TEAM_ID", "STARTING","PLAYER_IS_TEAM_STAR", "TEAM_STAR_OUT", 'NUM_USUAL_STARTERS_PRESENT', 'HOME_GAME']
    available_cols = [c for c in context_cols if c in test_df.columns]
    residuals_df = residuals_df.join(test_df[available_cols])
    
    return dict(RMSE=rmse, MAE=mae, R2=r2), residuals_df
```

```python
train_df = pd.concat([s21, s22, s23])
val_df = s24
test_df = s25

catModel, val_metrics, used_cats = fit_train_val(
    train_df, val_df,
    features=featuresV2,
    target_col="PTS",
    date_col="GAME_DATE",
    player_col="PLAYER_ID",
    use_gpu=False
)

print("Validation metrics:", val_metrics)

test_metrics, residuals_df = evaluate_test(catModel, test_df, featuresV2, "PTS", used_cats)
print("Test metrics:", test_metrics)
```
